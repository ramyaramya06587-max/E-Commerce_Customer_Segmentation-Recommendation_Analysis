
# E-Commerce Customer Segmentation & Recommendation Analysis

## Project Description
This project analyzes customer purchasing behavior using Machine Learning.

It uses K-Means clustering to identify meaningful customer segments and Cosine Similarity to recommend products.

## Objectives
- Analyze customer purchasing behavior.
- Perform RFM analysis.
- Segment customers using K-Means.
- Evaluate clusters using Elbow Method and Silhouette Analysis.
- Build a basic product recommendation system.
- Generate actionable business recommendations.
